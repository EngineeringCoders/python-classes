# write a function which takes student info and returns the student marks


````python
# variable keyword argument
# **kwargs
# in variable keyword argument the arguments behave as dictionary
def get_std_info(**std_details):
    # print(f"student details is : {std_details}")
    for k,v in std_details.items():
        print(f"{k}:{v}")
````

````python
get_std_info(name="Abhishek",roll_no="A101",course="AI")
````

````python
# write a function which takes either positional-only or keyword-only arguments
# use / to make parameters position-only. this is usefull 
# use * to make the parameters after it as keyword argument only
def ratio(numerator,denominator,/,*,as_percent=False):
    value = numerator/denominator
    return value * 100 if as_percent else value
````

````python
print(ratio(1,4))
````

````python
print(ratio(1,True))
````

````python
print(ratio(1,4,as_percent=True))
````

````python
# Default values are evaluated once when the function is define , not each time the function is called .
# Avoid mutable defaults like [],{} or set() unless we deliberatly want the shared state.
def append_bad(value,bucket=[]):
    bucket.append(value)
    return bucket

print(append_bad('first'))
print(append_bad('second'))
````

````python
def append_good(value,bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(value)
    return bucket

print(append_good('first'))
print(append_good('second'))
````

# create a function which takes input as 
def demo(*amounts,tax_rate=0.10,discount=0)
it should :
1. sum all the amounts
2. subtract the discount
3. apply tax
4. return the final value
5. print error if no amounts are passed 

````python
def price_function(*amounts, tax_rate=0.05, discount=0):
    if not amounts:
        print("No amounts provided. Please provide at least one amount.")
    total_amount = sum(amounts)
    tax_amount = total_amount * tax_rate
    final_amount = total_amount + tax_amount
    return final_amount

CONST_VALUE = 1 # strict instruction

print(f"Final Amount: ${price_function(100,200,300):.2f}");
print(f"Final Amount: ${price_function()}");
````

# higher order function : a function which accepts a function , returns a function or both 
# function is considered as first class object 

# In python , with function 
1. assign function to a variable .
2. store them in data structure .
3. pass them into other functions.
4. return them from other function.

````python
def square(x):
    return x*x
````

````python
# assign a function to a variable 
operation = square
print(operation)
````

````python
print(operation(2))
````

````python
operations = {
    'square': square,
    'cube': lambda x : x ** 3,
    'double': lambda x : x * 2
}
````

````python
for name, func in operations.items():
    print(name,func(2))
````
