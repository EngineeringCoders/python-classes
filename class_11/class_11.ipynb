```python
mistakes = [
    "syntax error",
    "indentation error",
    "syntax error",
    "name error",
    "indentation error",
    "indentation error"
]
```

```python
# find the frequency count of a given mistake we make while coding 
mistake_count = {}

for mistake in mistakes:
    if mistake not in mistake_count:
        mistake_count[mistake] = 0
    mistake_count[mistake] += 1
```

```python
print(mistake_count)
```

```python
from collections import Counter

mistake_count = Counter(mistakes)
print(mistake_count)
```

```python
# question - 2
attendance = [
    "Amit", "Neha", "Amit", "Ravi", "Neha",
    "Amit", "Sara", "Ravi", "Amit"
]

# How many times each student attended ?
# Find out the Top 2 most regular students in the calss ? 

from collections import Counter
import collections

attendance = [
    "Amit", "Neha", "Amit", "Ravi", "Neha",
    "Amit", "Sara", "Ravi", "Amit"
]

counts = Counter(attendance)
print(counts)
print(counts.most_common(2))
print(type(collections.Counter))

```

```python
students = [
    ("Shreya", "Python"),
    ("Abhishek","Data Science"),
    ("Sagar","Python"),
    ("Nishtha","Python"),
    ("Rusrith","AI")
]
# write a program to group students based on course they are enrolled into 
students = [
    ("Shreya", "Python"),
    ("Abhishek","Data Science"),
    ("Sagar","Python"),
    ("Nishtha","Python"),
    ("Rusrith","AI")
]

```

```python
# solution using traditional python logic ( Shivraj )

course_students = {}

for student, course in students:
    if course not in course_students:
        course_students[course] = []
    course_students[course].append(student) 

print(course_students)
```

```python
# using defaultdict of collections module
course_students_dict = defaultdict(list)
for student, course in students:
    course_students_dict[course].append(student)
print(course_students_dict)
```

```python
for course, names in course_students_dict.items():
    print(course, ":",names)
```

```python
# abhishek daya 

course_grp = defaultdict(list)
for name, course in students:
    course_grp[course].append(name)

for name, course in course_grp.items():
    print(name,":",course)
```

```python
# Recent Activity Tracker 
from collections import deque

recent_activities = deque(maxlen=3)

recent_activities.append("Amit submitted assignment")
recent_activities.append("Neha joined Python class")
recent_activities.append("Ravi completed quiz")
recent_activities.append("Sara asked a doubt")

print(recent_activities)
```
