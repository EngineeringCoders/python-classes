# reduce() function

```python
# this higher order funciton is used to reduce a given sequence of elements into a single element by applying a specified function
# reduce(function,sequence)
```

```python
# example 
from functools import reduce
ex_list = [10,20,30,40,50]
sum_result = reduce(lambda x,r: x+r,ex_list)
print(sum_result)
```

```python
# A function which takes input as an function as an argument 
# this special function is called a decorator function 
def decor(func):
    # extending the functionality of the input function (func)
    def inner(name):
        if name=="Abhishek":
            print(f"Hello {name}, Bad Morning!")
        else:
            func(name)
        
    # return inner function
    return inner
```

```python
@decor
def wish(name):
    print(f"Hello {name} Good Morning!")
```

```python
wish("Daya")
```

```python
wish("Abhishek")
```

```python
# call a decorator function by using below syntax 
result_of_decor_func = decor(wish)
```

```python
result_of_decor_func("Labiba")
```

```python
result_of_decor_func("Abhishek")
```

```python
# create a decorator function to extend functionality of input function 

def logger(func):
    def wrapper(*args,**kwargs):
        print("function started : ", func.__name__)
        result = func(*args,**kwargs)
        print("function ended:", func.__name__)
        return result
    return wrapper
```

```python
# report generation 
@logger
def generate_report(student):
    print("Name:",student["name"])
    print("Marks:",student["scores"])
    print("Attendance:", student["attendance"])
```

```python
generate_report(students[0])
```

```python
# Write a decorator function which gives us the execution time of a decorated function 
import time

def timer(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        print(f"{func.__name__} started at {start}")
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} ended on {end}")
        print(f"The total execution time of the function {func.__name__} is {time.localtime(end - start)}")

        return result
    return wrapper
```

```python
@timer
def students_report(students):
    for i in students:
        return generate_report(i)
```

```python
students_report(students)
```
