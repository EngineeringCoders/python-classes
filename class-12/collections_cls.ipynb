```python
# deque
from collections import deque
import timeit
```

```python
def list_queue():
    # FIFO
    items = list(range(10000))
    while items:
        items.pop(0)
```

```python
def deque_queue():
    # pop values from either of the side
    items = deque(range(10000))
    while items:
        items.popleft()
```

```python
lst_extraction_time = timeit.timeit(list_queue,number=10)
deque_extraction_time = timeit.timeit(deque_queue,number=10)
```

```python
print("List pop(o): ", lst_extraction_time)
print("Deque popleft(): ", deque_extraction_time)
```

```python
# ChainMap : It combines multiple dictionaries into one view.
# ChainMap searches from left to right in search of value of the accessed key. The first dictionary has the highest priority 
from collections import ChainMap

default_config = {
    "theme" : "light",
    "language": "English",
    "notificaitons": True
}

environment_config = {
    "language" : "Hindi"
}

user_config = {
    "theme" : "dark"
}
```

```python
# use of ChainMap

config = ChainMap(user_config, environment_config, default_config)
```

```python
print(config["theme"])
```

```python
print(config["notificaitons"])
```

## UserDict, UserList, UserString

- These are wrapper classes provided by collections module in python . 
- These are provided in order to create our own version of 
    - list
    - dict
    - str


```python
# create a custom dictionary data type that blocks negative marks
from collections import UserDict

class MarksDict(UserDict):
    def __setitem__(self, key, value):
        if value < 0:
            raise ValueError("Marks cannot be negative")
        self.data[key] = value
        
```

```python
marks = MarksDict()

marks["Abhishek"] = 99
marks["Hitasha"] = 89
```

```python
print(marks)
```

```python
print(type(marks))
```

```python
marks["Labiba"] = -99
```

```python
# Create a custom list that only accepts positive numbers
from collections import UserList

class CustomList(UserList):
    def __lt__(self, inp_item):
        if inp_item <= 0:
            raise ValueError("Negative item is not allowed!")
        self.data.append(inp_item)
    


list = CustomList()
print(list)
```
