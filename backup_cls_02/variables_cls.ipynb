```python
# range  : It is used to store sequence of numerical integer values with equal interval
# we can perform operation of indexing and slicing on range object 
# an object of range data type is always immutable in nature 


# syntax 
# var_range_a = range(6)
var_range_a = range(6)
print(var_range_a)
```

```python
for ele in var_range_a:
    print(ele)
```

```python
# 2nd way to declare range 
# range(start_ele, stop_ele)
var_range_b = range(11,21)
for ele in var_range_b:
    print(ele)
```

```python
# 3rd type of range declaration 
# range(start_ele, stop_ele,step)
var_range_b = range(11,21,3)
for ele in var_range_b:
    print(ele)
```
