```python
# returning multiple values from a given function 
# write a function which takes two input a,b returns sum, difference and product of a and b 

def calculate(a,b):
    """
    Returns the sum, difference and product of input numbers.

    Parameters:
        a: first input integer
        b: second input integer

    returns:
        a tuple of a+b, a-b, a*b
    """
    return a+b, a-b, a*b

add, sub, mul = calculate(2,3)
print(add,sub,mul)
```

```python
# we shouldn't start function name with digit
# we shouldn't make use of keywords as function name like if , for, return 
print(calculate.__doc__)
```

```python
# Variable Scope
# Local Scope 

message=""

def show():
    # local scope 
    message = "inside the function"
    print(message)

show()
# global scope 
print(message)
```

```python
# global scope 
count = 0

def increment():
    # accessing the global variable by using global keyword
    global count
    count += 1
```

```python
increment()
print(count)
```

```python
# better approach

def increment(count):
    return count + 1
increment(0)
```

```python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print(x)
        print(__doc__)
    inner()

outer()
```

```python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        """
        Returns the inner local variable
        """
        x = "local"
        print(x)
        print(inner.__doc__)
    inner()

outer()
```

```python
# nested function : use of nonlocal keyword

def counter():
    count = 0
    def increment():
        # nonlocal keyword tries to access the closest nonlocal variable with the same name 
        nonlocal count
        count += 1
        return count
    return increment

next_count = counter()
print(next_count)
# pass a call
print(next_count())
print(next_count())
```
